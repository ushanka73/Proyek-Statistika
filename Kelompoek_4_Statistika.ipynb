```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# ==========================================
# 1. LOAD DATASET
# ==========================================
# Catatan: Pastikan kamu sudah mengunggah file CSV datamu ke Google Colab/Jupyter
nama_file = "/content/Data Gabungan Kepadatan dan Kendaraan 2022 - 2025 Tanpa DOB Papua.xlsx"
df = pd.read_excel(nama_file)

print("===== DATA PENELITIAN BERHASIL DI-LOAD =====")
print(f"Total Sampel (N) : {len(df)} baris data (34 Provinsi x 4 Tahun)")
print(df[['Tahun', 'Provinsi', 'Kepadatan Penduduk', 'Jumlah Kendaraan']].head())
print("\n")

# ==========================================
# 2. STATISTIK DESKRIPTIF (Tahap 6 RTM)
# ==========================================
print("===== STATISTIK DESKRIPTIF =====")
deskriptif = df[['Kepadatan Penduduk', 'Jumlah Kendaraan']].describe()
print(deskriptif.loc[['mean', 'min', 'max', 'std']])
print("\n")

# ==========================================
# 3. ANALISIS REGRESI & PERHITUNGAN RESIDUAL
# ==========================================
# Menghitung persamaan regresi Y = a + bX
X = df['Kepadatan Penduduk']
Y = df['Jumlah Kendaraan']
slope, intercept, r_value, p_value, std_err = stats.linregress(X, Y)

# Menghitung prediksi dan residual untuk uji normalitas
df['Prediksi'] = intercept + (slope * X)
df['Residual'] = Y - df['Prediksi']

# ==========================================
# 4. UJI NORMALITAS RESIDUAL (Tahap 5 RTM)
# ==========================================
print("===== UJI NORMALITAS (SHAPIRO-WILK) =====")
stat_norm, p_norm = stats.shapiro(df['Residual'])
print(f"Shapiro-Wilk Statistic : {stat_norm:.4f}")
print(f"p-value Normalitas     : {p_norm:.5f}")
print("\n")
print("===== KEPUTUSAN UJI NORMALITAS =====")
if p_norm > 0.05:
    print("Karena p-value > 0.05, maka H0 DITERIMA.")
    print("Kesimpulan: Residual model berdistribusi normal (Syarat Regresi Terpenuhi).")
else:
    print("Karena p-value <= 0.05, maka H0 DITOLAK.")
    print("Kesimpulan: Residual tidak berdistribusi normal.")
    print("*(Catatan Akademik: Mengacu pada Central Limit Theorem, karena N=136 (>30),")
    print(" maka asumsi normalitas secara asimtotik tetap terpenuhi secara sah).*")
print("\n")

# ==========================================
# 5. HASIL UJI REGRESI LINEAR (Tahap 6 RTM)
# ==========================================
r_squared = r_value ** 2
t_statistic = slope / std_err

print("===== HASIL UJI REGRESI LINEAR =====")
print(f"R-Square (R2)            : {r_squared:.4f} ({r_squared*100:.2f}%)")
print(f"Konstanta (a)            : {intercept:.2f}")
print(f"Koefisien/Slope X (b)    : {slope:.2f}")
print(f"t-statistic              : {t_statistic:.3f}")
print(f"p-value Regresi (Sig.)   : {p_value:.5f}")
print("\n")

# ==========================================
# 6. KEPUTUSAN HIPOTESIS UTAMA (Tahap 6 RTM)
# ==========================================
print("===== KEPUTUSAN HIPOTESIS UTAMA =====")
if p_value < 0.05:
    print("Karena p-value < 0.05, maka H0 DITOLAK dan H1 DITERIMA.")
    print("Kesimpulan: Kepadatan Penduduk (X) BERPENGARUH SIGNIFIKAN terhadap Jumlah Kendaraan (Y).")
else:
    print("Karena p-value >= 0.05, maka H0 DITERIMA.")
    print("Kesimpulan: Kepadatan Penduduk (X) TIDAK BERPENGARUH signifikan terhadap Jumlah Kendaraan (Y).")
print("\n")

# ==========================================
# 7. INTERPRETASI OTOMATIS (Tahap 7 RTM)
# ==========================================
print("===== INTERPRETASI HASIL =====")
print(f"1. Persamaan regresi yang terbentuk adalah: Y = {intercept:.2f} + {slope:.2f} * X")
print(f"2. Nilai R-Square sebesar {r_squared*100:.2f}% menunjukkan bahwa Kepadatan Penduduk")
print(f"   mampu menjelaskan variasi Jumlah Kendaraan sebesar {r_squared*100:.2f}%.")
print(f"   Sedangkan {(1-r_squared)*100:.2f}% sisanya dijelaskan oleh faktor lain di luar penelitian.")

if slope > 0:
    print(f"3. Koefisien regresi bernilai positif ({slope:.2f}). Artinya, setiap kenaikan 1 jiwa/km²")
    print(f"   kepadatan penduduk, maka jumlah kendaraan diprediksi akan meningkat sebanyak {slope:.2f} unit.")
    print("   Fenomena nyata: Wilayah yang semakin padat memicu lonjakan kepemilikan kendaraan pribadi")
    print("   akibat mobilitas yang tinggi, yang pada akhirnya menjadi pemicu utama kemacetan lalu lintas.")
else:
    print(f"3. Koefisien regresi bernilai negatif ({slope:.2f}). Artinya, hubungan kedua variabel berbanding terbalik.")
print("\n")

# ==========================================
# 8. VISUALISASI GRAFIK
# ==========================================
print("===== GENERATING GRAPHICS... =====")
plt.figure(figsize=(12, 5))

# Grafik 1: Scatter Plot & Garis Regresi
plt.subplot(1, 2, 1)
sns.scatterplot(x=X, y=Y, color='blue', alpha=0.7, label='Data Riil')
# Membuat garis tren regresi
x_line = np.linspace(X.min(), X.max(), 100)
y_line = intercept + slope * x_line
plt.plot(x_line, y_line, color='red', linewidth=2, label='Garis Regresi')
plt.title('Scatter Plot & Garis Regresi Linear', fontsize=12)
plt.xlabel('Kepadatan Penduduk (jiwa/km²)')
plt.ylabel('Jumlah Kendaraan (unit)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Grafik 2: Histogram Distribusi Residual (Normalitas)
plt.subplot(1, 2, 2)
sns.histplot(df['Residual'], kde=True, color='green', bins=15)
plt.title('Grafik Distribusi Sebaran Residual', fontsize=12)
plt.xlabel('Nilai Residual')
plt.ylabel('Frekuensi')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
